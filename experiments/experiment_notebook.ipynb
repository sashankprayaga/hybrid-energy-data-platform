```python
# 1. Load the hybrid vehicle
vehicle = fastsim.Vehicle.from_resource(
    "2021_Hyundai_Sonata_Hybrid_Blue_thrml.yaml"
)

# 2. Load the UDDS driving cycle
cycle = fastsim.Cycle.from_resource("udds.csv")

# 3. Initialize the simulation
sim = fastsim.SimDrive(vehicle, cycle)

print("Simulation initialized successfully!")

# 4. Run the baseline simulation
sim.run()

print("Baseline simulation completed!")

# 5. Inspect available simulation outputs
print("\nSimulation output keys:")
print(sim.to_dict().keys())
```

```python
data = sim.to_dict()

hev = data["veh"]["pt_type"]["HEV"]

battery = hev["res"]["history"]
engine = hev["fc"]["history"]
motor = hev["em"]["history"]

# Battery state of charge
soc = battery["soc"]

# Engine fuel energy
fuel_energy = engine["energy_fuel_joules"]

# Engine and motor propulsion power
engine_power = engine["pwr_prop_watts"]
motor_power = motor["pwr_mech_prop_out_watts"]

print("\n=== HYBRID BASELINE ENERGY METRICS ===")

print("Battery SOC records:", len(soc))
print("Initial SOC:", soc[0])
print("Final SOC:", soc[-1])
print("SOC change:", soc[-1] - soc[0])

print("\nFuel energy records:", len(fuel_energy))
print("Fuel energy first:", fuel_energy[0])
print("Fuel energy last:", fuel_energy[-1])

print("\nMaximum engine propulsion power (kW):",
      max(engine_power) / 1000)

print("Maximum motor propulsion power (kW):",
      max(motor_power) / 1000)
```

```python
fuel_energy = engine["energy_fuel_joules"]

print("\n=== FUEL ENERGY VALIDATION ===")

print("First 10 values:")
print(fuel_energy[:10])

print("\nLast 10 values:")
print(fuel_energy[-10:])

print("\nMinimum value:", min(fuel_energy))
print("Maximum value:", max(fuel_energy))

print("\nSum of all values:", sum(fuel_energy))
print("Final value:", fuel_energy[-1])

print("\nNumber of decreases:",
      sum(
          fuel_energy[i] < fuel_energy[i - 1]
          for i in range(1, len(fuel_energy))
      ))
```

```python
data = sim.to_dict()
hev = data["veh"]["pt_type"]["HEV"]

battery = hev["res"]
soc = battery["history"]["soc"]

min_soc_allowed = battery["min_soc"]
max_soc_allowed = battery["max_soc"]

print("\n=== BATTERY SOC VALIDATION ===")

print("Configured minimum SOC:", min_soc_allowed)
print("Configured maximum SOC:", max_soc_allowed)

print("Observed minimum SOC:", min(soc))
print("Observed maximum SOC:", max(soc))

print("Initial SOC:", soc[0])
print("Final SOC:", soc[-1])

violations = sum(
    value < min_soc_allowed or value > max_soc_allowed
    for value in soc
)

print("SOC limit violations:", violations)
print("SOC validation passed:", violations == 0)
```

```python
import fastsim

vehicle = fastsim.Vehicle.from_resource(
    "2021_Hyundai_Sonata_Hybrid_Blue_thrml.yaml"
)
cycle = fastsim.Cycle.from_resource("udds.csv")

def inspect_soc(label, vehicle):
    hev = vehicle.to_dict()["pt_type"]["HEV"]
    battery = hev["res"]

    print(f"\n{label}")
    print("Battery state:", battery["state"])
    print("SOC balancing iterations:", hev["soc_bal_iters"])

inspect_soc("BEFORE simulation", vehicle)

sim = fastsim.SimDrive(vehicle, cycle)
sim.run()

result_vehicle = sim.to_dict()["veh"]
result_hev = result_vehicle["pt_type"]["HEV"]

print("\nAFTER simulation")
print("Battery state:", result_hev["res"]["state"])
print("First recorded SOC:", result_hev["res"]["history"]["soc"][0])
print("Last recorded SOC:", result_hev["res"]["history"]["soc"][-1])
print("SOC balancing iterations:", result_hev["soc_bal_iters"])
```

```python
import json
from pathlib import Path

data = sim.to_dict()

veh = data["veh"]
hev = veh["pt_type"]["HEV"]

soc = hev["res"]["history"]["soc"]
fuel_energy = hev["fc"]["history"]["energy_fuel_joules"]
vehicle_history = veh["history"]

summary = {
    "vehicle": veh["name"],
    "driving_cycle": "udds.csv",
    "fastsim_version": "3.1.0",
    "simulation_steps": len(vehicle_history["time_seconds"]),
    "distance_km": vehicle_history["dist_meters"][-1] / 1000,
    "fuel_energy_mj": fuel_energy[-1] / 1_000_000,
    "initial_recorded_soc": soc[0],
    "final_soc": soc[-1],
    "soc_change": soc[-1] - soc[0],
    "minimum_observed_soc": min(soc),
    "maximum_observed_soc": max(soc),
    "soc_limit_violations": sum(
        value < hev["res"]["min_soc"]
        or value > hev["res"]["max_soc"]
        for value in soc
    ),
    "cycle_tracking_failures": vehicle_history["cyc_met"].count(False),
    "soc_balancing_iterations": hev["soc_bal_iters"],
}

output_path = Path("results/baseline_summary.json")
output_path.parent.mkdir(parents=True, exist_ok=True)

with output_path.open("w") as f:
    json.dump(summary, f, indent=2)

print(f"\nBaseline summary saved to: {output_path}")
print(json.dumps(summary, indent=2))
```

```python

vehicle = fastsim.Vehicle.from_resource(
    "2021_Hyundai_Sonata_Hybrid_Blue_thrml.yaml"
)

cycle = fastsim.Cycle.from_resource("udds.csv")

sim = fastsim.SimDrive(vehicle, cycle)
sim.run()

data = sim.to_dict()

veh = data["veh"]
hev = veh["pt_type"]["HEV"]

histories = {
    "vehicle": veh["history"],
    "battery": hev["res"]["history"],
    "engine": hev["fc"]["history"],
    "motor": hev["em"]["history"],
}

print("\n=== TELEMETRY ALIGNMENT CHECK ===")

for name, history in histories.items():
    print(
        f"{name}: "
        f"{len(history['i'])} records"
    )

record_counts = {
    name: len(history["i"])
    for name, history in histories.items()
}

print("\nAll record counts match:",
      len(set(record_counts.values())) == 1)
```
